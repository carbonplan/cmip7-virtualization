# Search different catalogs

Check number of CMIP7 and CMIP6 datasets on each catalog endpoint

```python
from cmip7_virtualization.catalog import STAC_BASES, collection_counts
```

```python
for env, sites in STAC_BASES.items():
    for site, endpoints in sites.items():
        for base in endpoints["read"]:
            print(f"\n[{env}/{site}] {base}")
            try:
                for cid, n in collection_counts(base).items():
                    print(f"    {cid:12s} {n if n is not None else '?'}")
            except Exception as e:
                print(f"    ERROR: {type(e).__name__}: {e}")
```
